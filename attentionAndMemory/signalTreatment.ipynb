```python


def read_xdf_file(filepath):
    # Lire le fichier XDF
    streams, header = pyxdf.load_xdf(filepath)
    # Trouver le flux EEG (souvent il a un 'type' ou 'name' spécifique, ajustez si nécessaire)
    eeg_stream = next((stream for stream in streams if stream['info']['name'][0] == 'EEG'), None)
    if not eeg_stream:
        raise ValueError("No EEG stream found in the XDF file")
    
    # Récupérer les données EEG
    eeg_data = eeg_stream['time_series']
    # Récupérer les timestamps
    timestamps = eeg_stream['time_stamps']
    return eeg_data, timestamps

eeg_data, timestamps = read_xdf_file('./signaux EEG/Kayssieh_Charles_Congruent_1.xdf')
```

```python
def save_segments(segments, save_dir, base_filename):
    for i, segment in enumerate(segments):
        segment_filename = os.path.join(save_dir, f"{base_filename}_segment_{i}.npy")
        np.save(segment_filename, segment)
    print(f"Saved {len(segments)} segments for {base_filename}")
```

```python
def process_xdf_files(directory, timecodes, save_dir='./processed'):
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)
    
    for subdir, dirs, files in os.walk(directory, topdown=True):
        for filename in files:
            if filename.endswith(".xdf"):
                filepath = os.path.join(subdir, filename)
                
                try:
                    eeg_data, timestamps = read_xdf_file(filepath)
                    print(f"Successfully opened: {filepath}")
                    
                    segments = segment_data(eeg_data, timestamps, timecodes)
                    save_segments(segments, save_dir, filename.split('.')[0])
                except Exception as e:
                    print(f"Failed to process {filepath}: {e}")
```
